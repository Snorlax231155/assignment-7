## Task 1: Data Understanding

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 1. Load the dataset using Pandas
df = pd.read_csv('Mall_Customers.csv')

# 2. Display the first five records
print("First five records of the dataset:")
display(df.head())

# 3. Identify Numerical features and Categorical features
numerical_features = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = df.select_dtypes(exclude=[np.number]).columns.tolist()
print(f"Numerical features: {numerical_features}")
print(f"Categorical features: {categorical_features}")

# 4. Display the dataset information and summary statistics
print("\nDataset Information:")
df.info()

print("\nSummary Statistics:")
display(df.describe(include='all'))
```

## Task 2: Data Preprocessing

```python
# Check for missing values
print("Missing values in each column:")
print(df.isnull().sum())

# Remove unnecessary columns (e.g., CustomerID)
df_cleaned = df.drop(columns=['CustomerID'])

# Encode categorical variables if required
le = LabelEncoder()
df_cleaned['Gender'] = le.fit_transform(df_cleaned['Gender'])
print(f"Gender encoding mapping: Female -> 0, Male -> 1")

# Standardize the features using StandardScaler
scaler = StandardScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df_cleaned), columns=df_cleaned.columns)
print("\nStandardized Features (first 5 rows):")
display(df_scaled.head())
```

## Task 3: Model Development

```python
# 1. Use the Elbow Method to determine the optimal number of clusters
inertia = []
K = range(1, 11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(df_scaled)
    inertia.append(kmeans.inertia_)

# We will plot the Elbow Curve in the next task.
# 2. Train a K-Means Clustering model using the selected value of K (K=5)
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
kmeans.fit(df_scaled)

# 3. Assign cluster labels to each customer
df['Cluster'] = kmeans.labels_
df_cleaned['Cluster'] = kmeans.labels_
df_scaled['Cluster'] = kmeans.labels_

# 4. Apply Principal Component Analysis (PCA) to reduce the dataset to 2 principal components
pca = PCA(n_components=2, random_state=42)
pca_features = pca.fit_transform(df_scaled.drop(columns=['Cluster']))
df['PCA1'] = pca_features[:, 0]
df['PCA2'] = pca_features[:, 1]

print("PCA reduction complete. Shape of PCA features:", pca_features.shape)
print("Explained variance ratio of first 2 components:", pca.explained_variance_ratio_)
```

## Task 4: Visualization and Evaluation

```python
# 1. Generate Elbow Curve
plt.figure(figsize=(8, 5))
plt.plot(K, inertia, 'bo-', markersize=8, linewidth=2)
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia (WCSS)')
plt.title('Elbow Method for Optimal K')
plt.grid(True)
plt.show()

# 2. Scatter plot showing customer clusters (Annual Income vs Spending Score)
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster', palette='viridis', s=100, alpha=0.8)
plt.title('Customer Segments (Annual Income vs Spending Score)')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend(title='Cluster')
plt.grid(True)
plt.show()

# 3. PCA visualization with cluster labels
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='PCA1', y='PCA2', hue='Cluster', palette='viridis', s=100, alpha=0.8)
plt.title('PCA 2D Visualization of Customer Clusters')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Cluster')
plt.grid(True)
plt.show()
```

### Task 4 Observations

1. **Optimal Number of Clusters**: According to the Elbow Curve, the inertia (within-cluster sum of squares) decreases rapidly as K increases from 1 to 5, and the rate of decrease slows down significantly after K=5 (the elbow). Thus, K=5 is chosen as the optimal number of clusters.
2. **PCA Representation**: Since our preprocessed data has 4 features (Gender, Age, Annual Income, and Spending Score), visualizing them in a single 2D plot is impossible. Principal Component Analysis (PCA) projects this 4D space into 2 orthogonal directions (Principal Components 1 and 2) that capture the maximum variance (~60% total variance). This allows us to visualize the cluster boundaries and spacing clearly in 2D.
3. **Cluster Characteristics**:
   - **Cluster 0 (High Income, High Spending)**: Mean Age ~32.7, Mean Income ~86.5k, Mean Spending Score ~82.1. Approximately equal mix of Males and Females (54% Female).
   - **Cluster 1 (High Income, Low Spending)**: Mean Age ~36.5, Mean Income ~89.5k, Mean Spending Score ~18.0. Approximately 55% Female.
   - **Cluster 2 (Older Females, Mid Income, Mid-to-Low Spending)**: Mean Age ~49.8, Mean Income ~49.2k, Mean Spending Score ~40.1. This cluster consists of 100% Female customers.
   - **Cluster 3 (Young, Low Income, High Spending)**: Mean Age ~24.9, Mean Income ~39.7k, Mean Spending Score ~61.2. Approximately 59% Female.
   - **Cluster 4 (Older Males, Mid Income, Mid-to-Low Spending)**: Mean Age ~55.7, Mean Income ~53.7k, Mean Spending Score ~36.8. This cluster consists of 100% Male customers.
